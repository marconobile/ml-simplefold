```python
count = 0
for idx, el in enumerate(l):
    data = np.load(el, allow_pickle=True)
    count += data['trajectory'].shape[0]
    if idx == 0:
        new_trajectory = data['trajectory']
        new_residue_cluster_ids = data['residue_cluster_ids']
        new_dihedrals = data['dihedrals']
        new_frame_state_ids = data['frame_state_ids']
    else:
        new_trajectory = np.concatenate((new_trajectory, data['trajectory']), axis=0)
        new_residue_cluster_ids = np.concatenate((new_residue_cluster_ids, data['residue_cluster_ids']), axis=0)
        new_dihedrals = np.concatenate((new_dihedrals, data['dihedrals']), axis=0)
        new_frame_state_ids = np.concatenate((new_frame_state_ids, data['frame_state_ids']), axis=0)
assert new_trajectory.shape[0] == count        
```

```python
d = {
    'trajectory' : new_trajectory,
    'residue_cluster_ids' : new_residue_cluster_ids,
    'dihedrals' : new_dihedrals,
    'frame_state_ids' : new_frame_state_ids,
    'frame_indices': np.arange(new_trajectory.shape[0])
}
```

```python
data = dict(data)
```

```python
for k, v in data.items():
    if k in d: continue
    d[k] = v
```
